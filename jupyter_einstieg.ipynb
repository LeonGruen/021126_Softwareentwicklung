# Jupyter: interaktiv mit Python arbeiten

Jupyter-Notebooks verbinden **Text**, **Python-Code** und dessen **Ergebnisse** in einer Datei. Das ist praktisch, wenn du Daten schrittweise untersuchen, Berechnungen ausprobieren oder Ergebnisse mit Erklärungen dokumentieren möchtest.

Ein Notebook besteht aus Zellen. Textzellen erklären etwas; Codezellen enthalten Python. Du kannst jede Codezelle einzeln ausführen. Dabei merkt sich der sogenannte **Kernel** Variablen, solange er läuft.

## So führst du das Beispiel aus

1. Klicke in die Codezelle unten.
2. Drücke **Shift+Enter**, um sie auszuführen und zur nächsten Zelle zu springen.
3. Ändere zum Ausprobieren die Temperaturwerte und führe die Codezelle erneut aus.
4. Falls Ergebnisse unerwartet sind, starte den Kernel neu und führe die Zellen danach der Reihe nach aus.

```python
# Temperaturen einer Woche in Grad Celsius
temperaturen = [20, 23, 18, 20, 19]

mittelwert = sum(temperaturen) / len(temperaturen)
kuehlster_tag = min(temperaturen)
waermster_tag = max(temperaturen)

print(f'Wochendurchschnitt: {mittelwert:.1f} °C')
print(f'Kühlster Tag: {kuehlster_tag} °C')
print(f'Wärmster Tag: {waermster_tag} °C')
```

```python
import matplotlib.pyplot as plt

tage = range(1, len(temperaturen) + 1)
plt.plot(tage, temperaturen, marker="o", linewidth=2)
plt.title("Temperaturverlauf einer Woche")
plt.xlabel("Tag")
plt.ylabel("Temperatur (°C)")
plt.xticks(tage, [f"Tag {tag}" for tag in tage])
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Ergebnis verstehen

Die Codezelle berechnet den Durchschnitt sowie den kleinsten und größten Wert der Liste. Mit `print(...)` werden die Ergebnisse direkt unter der Zelle angezeigt. Die Liste `temperaturen` bleibt im Kernel gespeichert und kann auch in weiteren Codezellen verwendet werden.
